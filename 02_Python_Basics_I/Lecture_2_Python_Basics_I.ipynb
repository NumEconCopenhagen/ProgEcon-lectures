## 2. Variables and basic types

A **variable** is a name that refers to a value. Every value has a **type**.

The four basic types are:

- `int`: whole numbers,
- `float`: decimal numbers,
- `str`: text, and
- `bool`: `True` or `False`.

```python
students = 120
coffee_price = 38.0
city = 'Copenhagen'
lecture_today = True

print(type(students))
print(type(coffee_price))
print(type(city))
print(type(lecture_today))
```

Python allows underscores inside large numbers. They make the number easier to read but do not change its value.

```python
population = 6_000_000
print(population)
```

Text must be written inside quotation marks. Single and double quotation marks both work.

```python
good_1 = 'rye bread'
good_2 = "milk"

print(good_1)
print(good_2)
```

We can convert between some types. This is useful when a number has been read as text.

```python
price_as_text = '24.5'
price = float(price_as_text)

print(price, type(price))
```

## 3. Arithmetic and comparisons

The main arithmetic operators are `+`, `-`, `*`, `/` and `**`.

```python
price = 24.0
quantity = 3

expenditure = price * quantity
print('Expenditure:', expenditure, 'DKK')
print('Price squared:', price**2)
```

Comparisons return a boolean value.

```python
budget = 100.0
expenditure = 72.0

affordable = expenditure <= budget
print(affordable, type(affordable))
```

The most common comparisons are:

| Meaning | Python |
| --- | --- |
| equal | `==` |
| not equal | `!=` |
| less than | `<` |
| less than or equal | `<=` |
| greater than | `>` |
| greater than or equal | `>=` |

```python
x = 3
y = 5

print(x == y)
print(x != y)
print(x < y)
```

<a id="lists"></a>
## 4. Lists

A **list** stores several values in an ordered sequence. Python counts from zero, so the first element has index `0`.

```python
prices = [24.0, 8.0, 45.0, 18.0]

print(prices)
print('Number of prices:', len(prices))
print('First price:', prices[0])
print('Last price:', prices[-1])
```

A **slice** extracts part of a list. The start is included and the end is excluded.

```python
print(prices[:2])
print(prices[2:])
```

Lists are mutable. Their elements can be changed, and new elements can be added with the method `.append()`.

```python
prices[0] = 25.0
prices.append(12.0)

print(prices)
```

<a id="dictionaries"></a>
## 5. Dictionaries

A **dictionary** stores values under names called keys. This is useful when the meaning of each number matters.

```python
prices = {
    'rye bread': 24.0,
    'milk': 8.0,
    'coffee': 45.0,
}

print(prices['milk'])
```

Values can be changed or added using their key.

```python
prices['milk'] = 8.5
prices['apples'] = 20.0

print(prices)
```

A dictionary can also store simple information about countries.

```python
capitals = {
    'Denmark': 'Copenhagen',
    'Sweden': 'Stockholm',
    'Norway': 'Oslo',
}

print(capitals['Sweden'])
```

## 6. Conditionals

An `if` statement lets the program choose what to do. Indentation is part of the Python syntax.

```python
income = 10_000.0
expenditure = 8_500.0

if expenditure <= income:
    print('The bundle is affordable.')
else:
    print('The bundle is too expensive.')
```

Use `elif` when there are more than two possibilities.

```python
growth = 0.018

if growth < 0:
    print('Output fell.')
elif growth > 0:
    print('Output increased.')
else:
    print('Output was unchanged.')
```

<a id="for-loops"></a>
## 7. For loops

A `for` loop repeats the same operation for each element in a container.

```python
prices = [24.0, 8.0, 45.0]

for price in prices:
    print('Price including an illustrative 25 pct. tax:', 1.25 * price)
```

A loop can build a new list. Here we compute growth rates from an illustrative GDP index.

```python
gdp = [100.0, 103.0, 107.0, 110.0]
growth_rates = []

for t in range(len(gdp) - 1):
    growth = gdp[t+1] / gdp[t] - 1
    growth_rates.append(round(growth, 4))

print(growth_rates)
```

`range(3)` produces the integers `0`, `1` and `2`. The end point is not included.

```python
for i in range(3):
    print(i)
```
