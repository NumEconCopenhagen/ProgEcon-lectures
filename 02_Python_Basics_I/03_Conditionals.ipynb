## 1. <a id='toc1_'></a>[Conditionals](#toc0_)

You typically want your program to do one thing if some condition is met, and another thing if another condition is met. 

In Python this is done with **conditional statements**:

```python
x = 1
if x < 2: 
    # runs when x is smaller than 2
    print('first possibility')    
elif x > 4: # elif = else if
    # runs when x is not smaller than 2 and x is larger than 4
    print('second possibility')
elif x < 3:
    # runs when x is at least 2 and smaller than 3
    print('third possibility')
else:
    # runs when x is at least 3 and at most 4
    print('fourth possibility')  
```

<details>
<summary><strong>How Python chooses a branch</strong></summary>

Python checks the conditions from top to bottom. It runs the first branch whose condition is `True` and skips the remaining branches. The indentation shows which instructions belong to each branch.

A practical way to read the cell is to write down the value of `x`, evaluate each condition as `True` or `False`, and stop at the first `True` result.

</details>


**Try it yourself, 3 minutes:** Run the cell with `x = 1`, `x = 2.5` and `x = 5`. Before each run, predict which message will be printed.


**Note:**

1. "elif" is short for "else if" 
2. the **indentation** after if, elif and else is required (tabs or spaces, typically 4)

An **equivalent formulation** of the above if-elif-else statement is:

```python
x = 1
cond_1 = x < 2 # a boolean (True or False)
cond_2 = x > 4 # a boolean (True or False)
cond_3 = x < 3 # a boolean (True or False)
if cond_1: 
    print('first possibility')
elif cond_2:
    print('second possibility')
elif cond_3:
    print('third possibility')
else:
    print('fourth possibility')
```

The above can also be written purely in terms of if-statements:

```python
if cond_1: 
    print('first possibility')
if not cond_1 and cond_2:
    print('second possibility')
if not (cond_1 or cond_2) and cond_3:
    print('third possibility')
if not (cond_1 or cond_2 or cond_3):
    print('fourth possibility')
```

### 1.1. <a id='toc1_1_'></a>[True/False definitions of other types:](#toc0_)

If you place non-boolean types after the `if` statement, Python will transfer it to a boolean value by calling `bool()` on it.

This can be useful, but you should be careful.

```python
y = [1,2]
if y:
    print('y is not empty')
```

```python
print('bool([1,2]) = ',bool([1,2]))
print('bool([]) = ',bool([])) # Checks if list is empty
print('bool(0) = ',bool(0))
print('bool(5) = ',bool(5)) # This is true for all integers that are not 0 (also negative)
```

```python
# Don't do this with floats
print('bool(0.0) = ',bool(0.0))
print('bool(0.2) = ',bool(0.2))
close_to_zero = 1e-100
print('bool(1e-100) = ',bool(1e-100)) # True! Only exactly 0.0 is False -- 1e-100 is tiny but nonzero
```
