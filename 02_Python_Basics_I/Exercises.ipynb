## 6. Conditionals inside a loop: a simple tax schedule

Consider the following **fictional** tax schedule. Income is measured in thousand DKK.

- No tax on income up to 50.
- 20 percent tax on the part of income between 50 and 100.
- 40 percent tax on the part of income above 100.

This implies:

$$
\text{tax}(y) =
\begin{cases}
0 & \text{if } y \leq 50, \\
0.20(y-50) & \text{if } 50 < y \leq 100, \\
0.20 \cdot 50 + 0.40(y-100) & \text{if } y > 100.
\end{cases}
$$

Before writing code, calculate the tax for incomes 75 and 150 by hand.

Then:

1. Create an empty list called `taxes`.
2. Loop over `incomes`.
3. Use `if`, `elif` and `else` to calculate the tax.
4. Append each tax to `taxes`.
5. Compute after-tax income.
6. Print income, tax and after-tax income inside the loop.
7. Print the complete list of taxes after the loop.

````python
incomes = [40.0, 50.0, 75.0, 100.0, 150.0]
````

````python
# Write your code here
````

<details>
<summary><strong>Show hint</strong></summary>

Use the conditions in this order:

```python
if income <= 50:
elif income <= 100:
else:
```

The second branch is reached only when the first condition was `False`, so it already implies that income is above 50.

</details>

<details>
<summary><strong>Reveal solution</strong></summary>

```python
taxes = []

for income in incomes:
    if income <= 50:
        tax = 0.0
    elif income <= 100:
        tax = 0.20 * (income - 50)
    else:
        tax = 0.20 * 50 + 0.40 * (income - 100)

    taxes.append(tax)
    after_tax_income = income - tax

    print('income =', income, ', tax =', tax, ', after-tax income =', after_tax_income)

print('taxes =', taxes)
```

The taxes are `[0.0, 0.0, 5.0, 10.0, 30.0]`.

</details>
