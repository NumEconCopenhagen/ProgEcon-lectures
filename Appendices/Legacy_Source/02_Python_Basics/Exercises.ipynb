## 1. <a id='toc1_'></a>[Types and type conversion](#toc0_)

3. Convert the boolean `True` to an integer and print it.

```python
# write your code here
```

**Answer:**

```python
x = 7
y = 2

# 1. three kinds of division
print(x/y, type(x/y))   # standard division -> always a float
print(x//y, type(x//y)) # floor division -> int when both are ints
print(x%y, type(x%y))   # remainder -> int when both are ints

# 2. string -> float
z = float('3.5')
print(x+z, type(x+z))   # int + float -> float

# 3. bool -> int
print(int(True))
```

**Check:** floor division and the remainder must add back up to the original number, i.e. $x = (x // y) \cdot y + (x \% y)$.

**Note:** `x/y` returns a float *even when the division is exact* (`4/2` is `2.0`, not `2`). This is a common source of surprise when you index a list with the result of a division: indices must be integers.

## 2. <a id='toc2_'></a>[A budget check with booleans](#toc0_)

**Task:** A consumer has income $I = 10$ and faces prices $p_1 = 1$ and $p_2 = 2$. She considers two bundles: $(4,3)$ and $(4,4)$.

1. Compute the total expenditure $p_1 x_1 + p_2 x_2$ for each bundle.
2. For each bundle, create a boolean variable that is `True` if the bundle satisfies the budget constraint $p_1 x_1 + p_2 x_2 \leq I$.
3. Print the type of the boolean and the amount of income left over in each case.

```python
# write your code here
```

**Answer:**

```python
I = 10
p1 = 1
p2 = 2

# bundle A
x1_A = 4
x2_A = 3
expenditure_A = p1*x1_A + p2*x2_A
affordable_A = expenditure_A <= I

print('bundle A: expenditure =', expenditure_A)
print('bundle A: affordable =', affordable_A, 'of type', type(affordable_A))
print('bundle A: income left =', I-expenditure_A)
print('')

# bundle B
x1_B = 4
x2_B = 4
expenditure_B = p1*x1_B + p2*x2_B
affordable_B = expenditure_B <= I

print('bundle B: expenditure =', expenditure_B)
print('bundle B: affordable =', affordable_B, 'of type', type(affordable_B))
print('bundle B: income left =', I-expenditure_B)
```

**Note:** bundle A exhausts the budget *exactly*, so it is affordable only because the constraint is written with `<=` and not `<`. For bundle B the income left is negative, and the bundle is not affordable.

A comparison always returns a `bool`, and a `bool` can be used directly in an `if` statement:

```python
for name, affordable in zip(['A','B'],[affordable_A,affordable_B]):
    if affordable:
        print(f'bundle {name} can be bought')
    else:
        print(f'bundle {name} is too expensive')
```

**Careful:** the boundary case only works out here because all the numbers are *integers*. Had the prices been decimal numbers, `expenditure <= I` could easily have been `False` for a bundle that exactly exhausts the budget on paper - see exercise 10.

## 3. <a id='toc3_'></a>[Lists and slicing: growth rates](#toc0_)

**Task:** GDP in five consecutive years is given by the list below.

1. Print the first three observations and the last two observations using **slicing**.
2. Create a list `growth` with the yearly growth rates $g_t = Y_{t+1}/Y_t - 1$.
3. Print the number of growth rates and check that it is one less than the number of GDP observations.

```python
gdp = [100.0, 104.0, 109.0, 115.0, 122.0]
```

```python
# write your code here
```

**Answer:**

```python
gdp = [100.0, 104.0, 109.0, 115.0, 122.0]

# 1. slicing
print('first three:', gdp[:3])
print('last two:', gdp[-2:])

# 2. growth rates (with a loop)
growth = []
for t in range(len(gdp)-1):
    growth.append(gdp[t+1]/gdp[t]-1)

print('growth =', growth)

# the same with a list comprehension
growth_alt = [gdp[t+1]/gdp[t]-1 for t in range(len(gdp)-1)]
print('same result:', growth == growth_alt)

# 3.
print('number of growth rates:', len(growth))
assert len(growth) == len(gdp)-1
```

**Check:** compounding all the growth rates must bring us from the first to the last observation.

```python
import math

Y = gdp[0]
for g in growth:
    Y *= (1+g)

print('compounded:', Y)
print('actual    :', gdp[-1])
print('difference:', Y-gdp[-1])
print('equal within tolerance:', math.isclose(Y, gdp[-1]))
```

**Note:** here the difference happens to be exactly zero, but do not count on that. The compounded value goes through eight floating point divisions and multiplications, and each of them can introduce a tiny error - change the numbers in `gdp` and the difference is often of the order $10^{-14}$ instead. That is why we compare with `math.isclose` and not with `==` (see exercise 10).

## 4. <a id='toc4_'></a>[Copying vs. referencing](#toc0_)

**Task:** Consider the code below.

1. **First predict**, without running it, what `baseline` is after the last line.
2. Then run it. Were you right?
3. Fix the code so that `baseline` is left unchanged.

```python
baseline = [100.0, 104.0, 109.0]

scenario = baseline     # a "copy" of the baseline?
scenario[0] = 200.0     # a shock in the first year

print('scenario =', scenario)
print('baseline =', baseline)
```

```python
# write your fix here
```

**Answer:**

`baseline` is changed as well. The `=` sign copies the **reference**, not the content: `scenario` and `baseline` are two names for the *same* list object.

```python
baseline = [100.0, 104.0, 109.0]
scenario = baseline

print('same object:', scenario is baseline) # True -> the bug
```

The fix is to make an actual copy:

```python
from copy import copy

baseline = [100.0, 104.0, 109.0]
scenario = copy(baseline) # a new list object with the same content
scenario[0] = 200.0

print('scenario =', scenario)
print('baseline =', baseline)
print('same object:', scenario is baseline) # False
print('same content:', scenario == baseline) # False, we changed the first element

assert baseline == [100.0, 104.0, 109.0]
print('check passed: baseline is untouched')
```

**Note:** this is one of the most frequent bugs in this course, and it is silent - nothing crashes, your numbers are just wrong. Whenever you want an independent copy of a container, say so explicitly with `copy()`.

## 5. <a id='toc5_'></a>[Dictionaries: an expenditure calculation](#toc0_)

**Task:** Prices and quantities for three goods are given in the two dictionaries below.

2. Compute the total expenditure.
3. Print the price of `'butter'` if it is in the dictionary, and an informative message if it is not.

```python
prices = {'bread': 20.0, 'milk': 8.0, 'coffee': 45.0}
quantities = {'bread': 3.0, 'milk': 5.0, 'coffee': 2.0}
```

```python
# write your code here
```

**Answer:**

```python
prices = {'bread': 20.0, 'milk': 8.0, 'coffee': 45.0}
quantities = {'bread': 3.0, 'milk': 5.0, 'coffee': 2.0}

# 1. and 2.
total = 0.0
for good, p in prices.items():
    q = quantities[good]
    print(f'{good}: {p} x {q} = {p*q}')
    total += p*q

print('total =', total)

# 3.
if 'butter' in prices:
    print('the price of butter is', prices['butter'])
else:
    print('butter is not in the dictionary')
```

**Check:** the same total computed in a different way - with `zip` over the values.

```python
total_alt = sum([p*q for p,q in zip(prices.values(),quantities.values())])
print('total_alt =', total_alt)
assert total == total_alt
print('check passed')
```

**Note:** the check with `zip` only works because the two dictionaries have their keys in the *same order*. Looking up `quantities[good]` by key, as in the loop above, is the safe way of doing it.

## 6. <a id='toc6_'></a>[Conditionals in a loop](#toc0_)

**Task:** For each income in the list below, compute the income tax with the following (fictional) schedule:

$$
\text{tax}(y) = \begin{cases}
0 & \text{if } y \leq 50 \\
0.3 \cdot (y-50) & \text{if } 50 < y \leq 150 \\
0.3 \cdot 100 + 0.5 \cdot (y-150) & \text{if } y > 150
\end{cases}
$$

Loop over the incomes, use `if`/`elif`/`else`, and collect the taxes in a list. Print income, tax and the average tax rate for each person.

```python
incomes = [30.0, 50.0, 100.0, 150.0, 300.0]
```

```python
# write your code here
```

**Answer:**

```python
incomes = [30.0, 50.0, 100.0, 150.0, 300.0]

taxes = []
for y in incomes:

    if y <= 50.0:
        tax = 0.0
    elif y <= 150.0:
        tax = 0.3*(y-50.0)
    else:
        tax = 0.3*100.0 + 0.5*(y-150.0)

    taxes.append(tax)
    print(f'income = {y:6.1f}, tax = {tax:6.1f}, average tax rate = {tax/y:.3f}')
```

**Check:** the tax schedule is continuous, so at the two kinks ($y=50$ and $y=150$) the two adjacent formulas must give the same tax.

```python
import math

print(math.isclose(0.0, 0.3*(50.0-50.0)))                     # at y = 50
print(math.isclose(0.3*(150.0-50.0), 0.3*100.0+0.5*(150.0-150.0))) # at y = 150
```

**Note:** the order of the conditions matters. Because `elif` is only reached when the previous condition was `False`, the condition `y <= 150` implicitly also means `y > 50`. If you reverse the order of the branches, everybody ends up in the first one.

## 7. <a id='toc7_'></a>[A while loop: doubling time](#toc0_)

**Task:** GDP is 100 and grows by 2 pct. per year.

1. Use a **while loop** to find the number of full years it takes before GDP has *at least* doubled.
2. Compare your answer with the analytical doubling time $\log(2)/\log(1+g)$.

**Hint:** the analytical doubling time is generally not a whole number - your loop should give the first *integer* year at which GDP has passed 200.

```python
# write your code here
```

**Answer:**

```python
import math

g = 0.02
Y0 = 100.0

Y = Y0
years = 0
while Y < 2*Y0:
    Y *= (1+g)
    years += 1

print('years (loop) =', years)
print('GDP after those years =', Y)

exact = math.log(2)/math.log(1+g)
print('analytical doubling time =', exact)
```

**Check:** the loop stops at the first *whole* year after the analytical doubling time, i.e. at `math.ceil(exact)`.

```python
assert years == math.ceil(exact)
print('check passed')

# and the year before, GDP had not yet doubled
print('GDP one year earlier =', Y0*(1+g)**(years-1))
```

**Note:** a `while` loop is the right tool here because we do not know beforehand how many iterations we need. Always make sure the condition can actually become `False` - otherwise the loop never ends (press the stop button in Jupyter if it happens).

**Answer:**

```python
def u_func(x1,x2,alpha=0.5):
    """Cobb-Douglas utility.

    Args:
        x1 (float): consumption of good 1
        x2 (float): consumption of good 2
        alpha (float): weight on good 1 (default 0.5)

    Returns:
        (float): utility

    """

    return x1**alpha * x2**(1-alpha)

def demand(I,p1,p2,alpha=0.5):
    """Optimal demand with Cobb-Douglas preferences.

    Args:
        I (float): income
        p1 (float): price of good 1
        p2 (float): price of good 2
        alpha (float): weight on good 1 (default 0.5)

    Returns:
        x1 (float): optimal consumption of good 1
        x2 (float): optimal consumption of good 2

    """

    x1 = alpha*I/p1
    x2 = (1-alpha)*I/p2

    return x1,x2 # a tuple

I = 10.0
p1 = 1.0
p2 = 2.0

x1,x2 = demand(I,p1,p2) # unpacking the tuple
print(f'x1 = {x1:.4f}, x2 = {x2:.4f}, u = {u_func(x1,x2):.4f}')

help(demand)
```

## 9. <a id='toc9_'></a>[Scope: a global variable trap](#toc0_)

**Task:** Look at the code below. **Predict what the two print statements produce** before running the cell. Then run it, and rewrite the function so the bug cannot happen.

```python
alpha = 0.5

def u(x1,x2):
    return x1**alpha * x2**(1-alpha) # alpha is NOT an argument

print(u(2.0,8.0))

alpha = 0.9 # somewhere else in a long notebook ...

print(u(2.0,8.0)) # the same call, a different result
```

```python
# write your improved version here
```

**Answer:**

The function reads `alpha` from the **global** scope, so its output changes when a variable *outside* the function is changed. The same call gives two different numbers.

The fix is to make `alpha` an argument, so everything the function needs is passed to it:

```python
def u(x1,x2,alpha=0.5):
    """Cobb-Douglas utility with alpha as an argument."""

    return x1**alpha * x2**(1-alpha)

alpha = 0.5
print(u(2.0,8.0))

alpha = 0.9 # this global variable is now irrelevant for the function
print(u(2.0,8.0))
```

```python
# the two calls now give exactly the same result
assert u(2.0,8.0) == u(2.0,8.0,alpha=0.5)
print('check passed: the function no longer depends on the global alpha')
```

**Note:** a function that only depends on its arguments can be tested, re-used and moved to a module. A function that reads global variables can break because of a cell you ran ten minutes ago, somewhere else in the notebook.

## 10. <a id='toc10_'></a>[Floating point numbers](#toc0_)

**Task:**

1. Start from `x = 0.0` and add `0.1` ten times in a loop. Print `x` with 20 decimals.
2. Test with `==` whether `x` equals `1.0`. Test the same with `math.isclose`.
3. Write a small function `equal(a,b)` that returns whether `a` and `b` are equal *within a tolerance*, and use it on `0.1+0.2` and `0.3`.

```python
# write your code here
```

**Answer:**

```python
import math

# 1.
x = 0.0
for _ in range(10):
    x += 0.1

print(f'x = {x:.20f}')

# 2.
print('x == 1.0        :', x == 1.0)
print('isclose(x,1.0)  :', math.isclose(x,1.0))

# 3.
def equal(a,b,tol=1e-8):
    """Return True if a and b are equal within the tolerance tol."""

    return abs(a-b) < tol

print('0.1+0.2 == 0.3  :', 0.1+0.2 == 0.3)
print('equal(0.1+0.2,0.3):', equal(0.1+0.2,0.3))
```

**Check:** how large is the error actually? Print the difference.

```python
print(f'{x-1.0:.20f}')
print(f'{(0.1+0.2)-0.3:.20f}')
```

**Note:** the errors are of the order $10^{-16}$ - completely irrelevant *economically*, but fatal if you use `==`. In the rest of the course, never test computed floats for exact equality: use `math.isclose` (or `np.isclose`/`np.allclose` for arrays).
