## 1. Functions

A **function** is a named block of code. It can take inputs, perform a calculation and return an output.

```python
def square(x):
    return x**2

print(square(4))
```

The indentation after `def` is required. The value after `return` becomes the output of the function.

```python
result = square(5)
print(result)
```

Functions are useful because a calculation only has to be written once.

```python
def growth_rate(old, new):
    return new / old - 1

print(round(growth_rate(100, 104), 4))
print(round(growth_rate(104, 109), 4))
```

<a id="arguments-outputs"></a>
## 2. Arguments and outputs

A function can have several arguments.

```python
def expenditure(price, quantity):
    return price * quantity

print(expenditure(24.0, 3))
```

A **keyword argument** has a default value. It can be changed when the function is called.

```python
def net_income(income, tax_rate=0.37):
    return income * (1 - tax_rate)

print(net_income(30_000))
print(net_income(30_000, tax_rate=0.25))
```

A short docstring explains what a function does.

```python
def real_value(nominal_value, price_index):
    """Convert a nominal value to a value at base-year prices."""
    return nominal_value / price_index * 100

print(real_value(525, 105))
```

A function can return two values. Python collects them in a tuple, which can be unpacked into two variables.

```python
def consumer_demand(income, p1, p2, alpha=0.5):
    x1 = alpha * income / p1
    x2 = (1 - alpha) * income / p2
    return x1, x2

x1, x2 = consumer_demand(1_000, 20, 40)
print('x1 =', x1)
print('x2 =', x2)
```

<a id="methods"></a>
## 3. Methods

A **method** is a function attached to an object. It is called with dot notation.

You have already used methods such as `.append()` for lists.

```python
goods = ['bread', 'milk']
goods.append('coffee')

print(goods)
```

Strings also have methods.

```python
city = 'copenhagen'
print(city.upper())
print(city.capitalize())
```

The distinction is simple:

- `len(goods)` is a function,
- `goods.append('coffee')` is a method of the list `goods`.

<a id="copies"></a>
## 4. References and copies

The assignment operator `=` does not make an independent copy of a list. It gives the same list a second name.

```python
baseline = [100.0, 104.0, 109.0]
scenario = baseline

scenario[0] = 200.0

print('scenario =', scenario)
print('baseline =', baseline)
```

Both variables changed because they refer to the same list. Use `.copy()` when the new list should be independent.

```python
baseline = [100.0, 104.0, 109.0]
scenario = baseline.copy()

scenario[0] = 200.0

print('scenario =', scenario)
print('baseline =', baseline)
```

`==` asks whether two objects contain the same values. `is` asks whether they are the same object.

```python
a = [1, 2]
b = [1, 2]
c = a

print(a == b, a is b)
print(a == c, a is c)
```

<a id="scope"></a>
## 5. Scope

Variables created inside a function are **local**. Variables created outside are **global**.

A function is easier to understand when everything it needs is passed as an argument.

```python
tax_rate = 0.37

def net_income_bad(income):
    return income * (1 - tax_rate)

print(net_income_bad(30_000))

tax_rate = 0.50
print(net_income_bad(30_000))
```

The same function call gave a different result because a global variable changed. Make the dependence explicit instead.

```python
def net_income(income, tax_rate=0.37):
    return income * (1 - tax_rate)

print(net_income(30_000))
print(net_income(30_000, tax_rate=0.50))
```

<a id="floating-point"></a>
## 6. Floating point numbers

Decimal numbers are stored approximately on a computer. Therefore, some calculations that are equal on paper are not exactly equal in Python.

```python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
```

Do not use `==` to compare numbers obtained from decimal calculations. Use `math.isclose()` instead.

```python
import math

print(math.isclose(0.1 + 0.2, 0.3))
```

The difference is extremely small, but exact equality still fails.

```python
difference = (0.1 + 0.2) - 0.3
print(f'{difference:.20f}')
```

### Exercise 1: A demand function

For Cobb-Douglas preferences, demand is

$$
x_1 = \alpha \frac{I}{p_1}, \qquad
x_2 = (1-\alpha)\frac{I}{p_2}.
$$

Write a function `demand(income, p1, p2, alpha=0.5)` that returns both quantities. Call it with `income=800`, `p1=20` and `p2=40`.

```python
# Write your code here
```

**One possible answer**

```python
def demand(income, p1, p2, alpha=0.5):
    x1 = alpha * income / p1
    x2 = (1 - alpha) * income / p2
    return x1, x2

x1, x2 = demand(800, 20, 40)
print(x1, x2)
```
