```python
def monthly_budget(net_income, rent, food=3000, transport=500):
    """Return total monthly expenses and money left."""

    total_expenses = rent + food + transport
    money_left = net_income - total_expenses

    return total_expenses, money_left

total_expenses, money_left = monthly_budget(18_000, 7_000)

print('total_expenses =', total_expenses)
print('money_left =', money_left)
print('identity holds =', 18_000 == total_expenses + money_left)

higher_expenses, lower_money_left = monthly_budget(
    18_000,
    7_000,
    food=4_000,
    transport=800
)

print('higher_expenses =', higher_expenses)
print('lower_money_left =', lower_money_left)
print('identity holds =', 18_000 == higher_expenses + lower_money_left)
```

With the default values, total expenses are 10,500 DKK and money left is 7,500 DKK.

With the higher food and transport expenses, total expenses are 11,800 DKK and money left is 6,200 DKK.

```python
class SavingsPlan:

    def __init__(self, name, initial_savings, monthly_saving):
        """Store the information for a savings plan."""

        self.name = name
        self.savings = initial_savings
        self.monthly_saving = monthly_saving

    def add_one_month(self):
        """Add one monthly saving amount."""

        self.savings = self.savings + self.monthly_saving

    def simulate(self, months):
        """Add savings for a given number of months."""

        for month in range(months):
            self.add_one_month()

anna_plan = SavingsPlan('Anna', 1000, 750)

print('name =', anna_plan.name)
print('initial savings =', anna_plan.savings)

anna_plan.add_one_month()
print('after one month =', anna_plan.savings)

anna_plan = SavingsPlan('Anna', 1000, 750)
anna_plan.simulate(6)
print('Anna after six months =', anna_plan.savings)

bo_plan = SavingsPlan('Bo', 2500, 600)
bo_plan.simulate(6)
print('Bo after six months =', bo_plan.savings)

print('Bo has more savings =', bo_plan.savings > anna_plan.savings)
```

Anna has 1,750 DKK after one month and 5,500 DKK after six months. Bo has 6,100 DKK after six months.

The class uses only three basic ideas:

- attributes store the data,
- methods perform operations on the data,
- `self` refers to the particular savings-plan object.

```python
class SavingsPlan:

    def __init__(self, name, initial_savings, monthly_saving):
        """Store the information for a savings plan."""

        self.name = name
        self.savings = initial_savings
        self.monthly_saving = monthly_saving

    def add_one_month(self):
        """Add one monthly saving amount."""

        self.savings = self.savings + self.monthly_saving

    def simulate(self, months):
        """Add savings for a given number of months."""

        for month in range(months):
            self.add_one_month()

    def months_to_target(self, target):
        """Return the number of months needed to reach a target."""

        future_savings = self.savings
        months = 0

        while future_savings < target:
            future_savings = future_savings + self.monthly_saving
            months = months + 1

        return months

plan = SavingsPlan('Anna', 1000, 750)

months = plan.months_to_target(5000)

print('months =', months)
print('current savings are unchanged =', plan.savings)
```

The target is reached after 6 months. `plan.savings` remains 1,000 DKK because the method changes only the local variable `future_savings`.
