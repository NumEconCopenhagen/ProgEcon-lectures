## 1. <a id='toc1_'></a>[Basic loops](#toc0_)

We often need to do the same task many times over. We use **loops** to do that. 

**Code repetition** gives you **horrible errors** and takes time.

**2 kinds of loop**
    
* `for` loop: when you know how many iterations to do
* `while` loop: when the stopping point is unknown

Never do this (can you spot the bug?):

```python
xs = [0,1,2,3,4]
x_sqr = []

x_sqr.append(xs[0]**2)
x_sqr.append(xs[1]**2)
x_sqr.append(xs[2]**2)
x_sqr.append(xs[4]**2)
print(x_sqr)
```

**Bug:** `xs[3]` was accidentally skipped — `xs[4]` was written instead. The output is `[0, 1, 4, 16]` rather than `[0, 1, 4, 9, 16]`. This kind of mistake is exactly what loops prevent.

Use a **for loop** instead:

```python
x_sqr = [] # empty list
for x in xs:
    x_sqr.append(x**2)
x_sqr
```

<details>
<summary><strong>Anatomy of the `for` loop</strong></summary>

- `x_sqr = []` creates an empty result list.
- `for x in xs:` assigns the next element of `xs` to `x` on each iteration.
- The indented line computes a square and appends it to the result.
- After the loop, every input element has been processed once.

The loop body is short, which makes it easier to inspect than five copied lines.

</details>


**Try it yourself, 4 minutes:** Given `incomes = [100,200,300]`, use a loop to create a new list containing income after a 20 percent tax. Print both lists.


Even better: a **list comprehension**. List comprehension is the shortest syntax and runs faster. Use that if you can.

```python
x_sqr = [x**2 for x in xs]
x_sqr
```

A **while loop**:

```python
i_sqr = [] # empty list
i = 0
while i < 6:
    i_sqr.append(i**2)
    i += 1
    
print(i_sqr)
```

Use a **for loop** with **range** instead:

```python
y_list = [] # empty list
for x in range(6):
    print(x)
    y_list.append(x**2)
print(y_list)
```

<details>
<summary><strong>Choosing between `for` and `while`</strong></summary>

Use `for` when you can describe the items or number of repetitions in advance. Use `while` when repetition should continue until a condition changes. In a `while` loop, check that at least one line moves the program toward stopping.

</details>


Range has some additional options, if more arguments are added: `range(start,stop,step)`

```python
for x in range(3,15,3): # Note that the range still ends 1 number before stop
    print(x)
```

## 2. <a id='toc2_'></a>[Nested loops](#toc0_)

```python
x_list = [1,2,3]
y_list = ['a','b','c']
for x in x_list:
    for y in y_list:
        print(x,y)
```

With `itertools` this can be written in one line:

```python
import itertools as it
for i,j in it.product(x_list,y_list):
    print(i,j)
```

## 3. <a id='toc3_'></a>[Enumerated loops](#toc0_)

```python
y_list = []
x_list = [10,25,31,40]

for i,x in enumerate(x_list):
    print('i =', i)
    y_list.append(x*i)
print('y_list =',y_list)
```

<details>
<summary><strong>What `enumerate` adds</strong></summary>

`enumerate(x_list)` produces both the position `i` and the value `x`. This is useful when the calculation needs the position as well as the element. If only the values are needed, a plain `for x in x_list` loop is simpler.

</details>


## 4. <a id='toc4_'></a>[`Continue` and `break`](#toc0_)

Loops can be fine-tuned with **continue** and **break**.

```python
y_list = []
x_list = [x for x in range(10)]

for i,x in enumerate(x_list):
    if i == 1:
        continue # go to next iteration 
    elif i == 4:
        break # stop loop prematurely
    y_list.append(x**2)
    
print(y_list)
```

<details>
<summary><strong>`continue` and `break`</strong></summary>

- `continue` skips the rest of the current iteration and moves to the next one.
- `break` ends the entire loop immediately.

Trace the example one value of `i` at a time. At `i = 1`, nothing is appended. At `i = 4`, the loop stops, so later values are never visited.

</details>


**Task:** Create a list with the 10 first positive uneven numbers.

```python
# write your code here
```

**Answer:**

```python
# The long way 
i = 1
x = []
while len(x)<10:
    
    if i%2==0: # The %-operator returns the remainder after division which is 0 for even numbers
        pass # Do nothing
    else:
        # Add the uneven number
        x.append(i)
    i +=1

print(x)
```

Note: The modulus operator % returns the remainder after division.

```python
print(3%2) # 3 divided by 2 leaves a remainder of 1
print(4%2) # 4 divided by 2 leaves a remainder of 0
print(5%3) # 5 divided by 3 leaves a remainder of 2
```

```python
# The pythonic way
end = 10*2+1 # The last number in the range is not included, so we need to add 1
my_list = [i for i in range(1,end,2)] # The third argument means we want every second number in the range from 1 to 21 
print(my_list)
```

```python
# The 'mathy' way
my_list = []
for i in range(1,11):
    my_list.append(i*2-1)
print(my_list)
```

## 5. <a id='toc5_'></a>[Zipped loops](#toc0_)

We can loop over **2 lists at the same time**:

```python
x = ['I', 'II', 'III']
y = ['a', 'b', 'c']

for i,j in zip(x,y):
    print(i+j)
```

**Try it yourself, 3 minutes:** Create lists `prices = [10,20,30]` and `quantities = [2,1,4]`. Use `zip` to loop over matching pairs and print each expenditure `price*quantity`.


Zipping is **not** the same as **nesting** 

```python
# A nested loop
for i in x:
    for j in y:
        print(i+j)
```

<details>
<summary><strong>`zip` versus a nested loop</strong></summary>

`zip(x,y)` forms matching pairs: first with first, second with second and so on. A nested loop forms every possible combination. With three elements in each list, `zip` gives 3 pairs while nesting gives 9 pairs.

</details>


## 6. <a id='toc6_'></a>[Dictionaries](#toc0_)

We can loop through keys, values or key-value pairs of a dictionary.

```python
my_dict = {'a': '-', 'b': '--', 'c': '---'}
for key in my_dict.keys():
    print(key)
```

```python
for key in my_dict: # default is to iterate over the keys of a dictionary
    print(key)
```

```python
for val in my_dict.values():
    print(val)
```

```python
for key,val in my_dict.items():
    print(key,val)
```

We can also **check whether a key exists**:

```python
if 'a' in my_dict:
    print('a is in my_dict with the value ' + my_dict['a'])
else:
    print('a is not in my_dict')
```

```python
if 'd' in my_dict:
    print('d is in my_dict with the value ' + my_dict['d'])
else:
    print('d is not in my_dict')
```
