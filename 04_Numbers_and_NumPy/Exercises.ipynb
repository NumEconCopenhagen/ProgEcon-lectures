Before running the code, predict whether the two comparisons will give the same answer.

````python
# Write your code here
````

<summary><strong>Show hint</strong></summary>

Begin with:

```python
total = 0.0

for repetition in range(10):
    total = total + 0.1
```

Use `math.isclose(total, 1.0)` for the comparison with a tolerance.

</details>

<details>
<summary><strong>Reveal solution</strong></summary>

```python
import math

total = 0.0

for repetition in range(10):
    total = total + 0.1

print('total =', total)
print('exactly equal to 1.0 =', total == 1.0)
print('close to 1.0 =', math.isclose(total, 1.0))
print('difference =', total - 1.0)
```

The result is close to:

```python
total = 0.9999999999999999
exactly equal to 1.0 = False
close to 1.0 = True
difference = -1.1102230246251565e-16
```

The small difference comes from the finite representation of decimal numbers. For computed floating-point numbers, use a comparison with a tolerance rather than `==`.

</details>

## 2. Prices and quantities as NumPy arrays

A household buys four goods. Prices and quantities are stored in NumPy arrays.

1. Print the type, shape, size and data type of `prices`.
2. Calculate expenditure on each good without using a loop:

$$
\text{item expenditure} = \text{price} \cdot \text{quantity}.
$$

3. Store the result as `item_expenditures` and print it.
4. Calculate and print total expenditure with `np.sum`.
5. Calculate and print mean expenditure per good with `np.mean`.
6. Print the expenditure on the first and last goods using indexing.
7. Create `higher_prices` by increasing every price by 5 percent.
8. Calculate total expenditure with the higher prices and print the change in total expenditure.

````python
import numpy as np

prices = np.array([18.0, 12.0, 30.0, 8.0])
quantities = np.array([2.0, 3.0, 1.0, 5.0])
````

````python
# Write your code here
````

<details>
<summary><strong>Show hint</strong></summary>

NumPy applies arithmetic element by element:

```python
item_expenditures = prices * quantities
```

A 5 percent price increase can be written as:

```python

```

</details>

<details>
<summary><strong>Reveal solution</strong></summary>

```python
print('type =', type(prices))
print('shape =', prices.shape)
print('size =', prices.size)

print('higher total expenditure =', higher_total_expenditure)
```

## 4. Select observations with logical indexing

The array below contains illustrative monthly incomes in DKK.

1. Calculate and print the mean income.
2. Create a boolean array called `above_mean` that is `True` when income is above the mean.
3. Use the boolean array to create `incomes_above_mean`.
4. Print the boolean array and the selected incomes.
5. Calculate the number of incomes above the mean using `.size`.
6. Calculate their share of all observations.
7. Create a second boolean array called `between_20_and_30` that selects incomes from 20,000 DKK to 30,000 DKK, including both limits.
8. Print the incomes selected by the second condition.

Use `&` to combine the two conditions in step 7.

````python
import numpy as np

monthly_incomes = np.array([
    18_000.0,
    24_000.0,
    32_000.0,
    15_000.0,
    41_000.0,
    26_000.0
])
````

````python
# Write your code here
````

<details>
<summary><strong>Show hint</strong></summary>

Logical indexing follows this pattern:

```python
condition = monthly_incomes > some_value
selected_values = monthly_incomes[condition]
```

Combine two array conditions with parentheses and `&`:

```python
(first_condition) & (second_condition)
```

</details>

<details>
<summary><strong>Reveal solution</strong></summary>

```python
mean_income = np.mean(monthly_incomes)

above_mean = monthly_incomes > mean_income
incomes_above_mean = monthly_incomes[above_mean]

number_above_mean = incomes_above_mean.size
share_above_mean = number_above_mean / monthly_incomes.size

between_20_and_30 = (
    (monthly_incomes >= 20_000) &
    (monthly_incomes <= 30_000)
)
incomes_between_20_and_30 = monthly_incomes[between_20_and_30]

print('mean income =', mean_income)
print('above mean =', above_mean)
print('incomes above mean =', incomes_above_mean)
print('number above mean =', number_above_mean)
print('share above mean =', share_above_mean)
print('incomes from 20,000 to 30,000 =',
      incomes_between_20_and_30)
```

The mean is 26,000 DKK. The incomes strictly above the mean are 32,000 DKK and 41,000 DKK. Their share is one third.

The incomes between 20,000 DKK and 30,000 DKK are 24,000 DKK and 26,000 DKK.

</details>

## 5. Optional: a two-dimensional sales table

A shop sells two products over three months.

- Each row in `quantities_sold` is a month.

- `prices` contains one price for each product.

1. Print the shape of both arrays.

5. Calculate total revenue for each product with `np.sum(..., axis=0)`.
6. Calculate overall revenue.
7. Use `math.isclose` to check that the sum of monthly revenue equals overall revenue.

````python
import math
import numpy as np

quantities_sold = np.array([
    [10.0, 5.0],
    [12.0, 4.0],
    [9.0, 7.0]
])

prices = np.array([25.0, 40.0])
````

````python
# Write your code here
````

<details>
<summary><strong>Show hint</strong></summary>

The one-dimensional price array matches the two columns of the quantities array:

```python
revenue = quantities_sold * prices
```

For this table:

- `axis=0` reduces the rows and leaves one value per column,
- `axis=1` reduces the columns and leaves one value per row.

</details>

<details>
<summary><strong>Reveal solution</strong></summary>

```python

print('prices shape =', prices.shape)

monthly_revenue = np.sum(revenue, axis=1)
product_revenue = np.sum(revenue, axis=0)
total_revenue = np.sum(revenue)

print('revenue table =')
print(revenue)
print('monthly revenue =', monthly_revenue)
print('product revenue =', product_revenue)
print('total revenue =', total_revenue)
print('totals agree =',
      math.isclose(np.sum(monthly_revenue), total_revenue))
```

