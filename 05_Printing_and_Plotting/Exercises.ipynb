```python
with open('revenue_report.txt', 'w') as report_file:

    report_file.write('Week    Price   Quantity    Revenue\n')

    total_revenue = 0

    for week, price, quantity in zip(weeks, prices, quantities):
        revenue = price * quantity
        total_revenue = total_revenue + revenue

        line = f'{week:4d} {price:8.2f} {quantity:10d} {revenue:10.2f}'
        report_file.write(line + '\n')

    report_file.write(f'Total revenue: {total_revenue:.2f} DKK\n')

with open('revenue_report.txt', 'r') as report_file:
    report = report_file.read()

print(report)
```

```python
labor_values = np.linspace(0, 10, 50)

output_grid = np.sqrt(capital_grid * labor_grid)

fig = plt.figure()
ax = fig.add_subplot(1, 1, 1, projection='3d')

ax.plot_surface(capital_grid, labor_grid, output_grid)

ax.set_title('Production surface')
ax.set_xlabel('Capital')
ax.set_ylabel('Labor')
ax.set_zlabel('Output')

fig.savefig('production_surface.png')
```
