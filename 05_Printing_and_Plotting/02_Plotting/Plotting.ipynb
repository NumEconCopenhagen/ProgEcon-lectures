## 1. <a id='toc1_'></a>[Line plots](#toc0_)


### 1.1. <a id='toc1_1_'></a>[Calculate a utility function on a vector](#toc0_)


Consider a Cobb-Douglas utility function

$$
u(x_1,x_2) = x_1^{\alpha}x_2^{1-\alpha}
$$

For now we fix $\alpha$ at 0.5, and $x_{2}$ at 1. And calculate utility across $x_{1}$.

```python
def u_func(x1,x2,alpha=0.50):
    return x1**alpha * x2**(1-alpha)
```

```python
# a. fixed value of x2 
x2 = 1.0

# b. vector of x1 from 0 to x_max with N values
x_max = 100.0
N = 1000
x1_vec = np.linspace(0,x_max,N)

# c. utility
u_vec = u_func(x1_vec,x2)
```

<details>
<summary><strong>Preparing data before plotting</strong></summary>

The cell first fixes a scalar value for $x_2$, then creates 1000 values for $x_1$ and finally evaluates utility for all 1000 values at once. The result `u_vec` has the same length as `x1_vec`, which is why the two arrays can be plotted against each other.

</details>


### 1.2. <a id='toc1_2_'></a>[Draw a line plot](#toc0_)


**Plot-object in matplotlib:**

1. `fig` is the figure object - it's the "canvas"
2. `ax` is a reference to a specific subplot on the canvas (only one here)

```python
fig = plt.figure()
ax = fig.add_subplot(1,1,1) # 1,1,1 is (nrows, ncols, index)
```

<details>
<summary><strong>`fig` and `ax` have different jobs</strong></summary>

`fig` is the entire figure. `ax` is the plotting area on that figure. Data, titles and axis labels are usually added through `ax`. Saving and arranging several plotting areas are usually handled through `fig`.

</details>


Add some content:

```python
fig = plt.figure()
ax = fig.add_subplot(1,1,1)

ax.plot(x1_vec,u_vec,ls='-',lw=2)

# ls is linestyle
# lw is line width

ax.set_title('Utility, $u(x_1,1)$')
ax.set_xlabel('$x_1$')
ax.set_ylabel('$u(x_1,1)$');
```

<details>
<summary><strong>The smallest useful line plot</strong></summary>

- `ax.plot(x1_vec,u_vec,...)` draws one line from paired values.
- `ls` controls line style and `lw` controls line width.
- `set_title`, `set_xlabel` and `set_ylabel` make the economic content readable.

This is enough for most first figures. Add further options only when they communicate something useful.

</details>

### 1.3. <a id='toc1_3_'></a>[Compare several curves](#toc0_)

Assume $x_{2}=2$ is also a possibility.


```python
x2_alt = 2.0
u_vec_2 = u_func(x1_vec,x2_alt)
```

Together:

```python
fig = plt.figure()
ax = fig.add_subplot(1,1,1)

ax.plot(x1_vec,u_vec,ls='-',lw='4',label='$x_{2}=1$')
ax.plot(x1_vec,u_vec_2,ls='--',lw='2',label='$x_{2}=2$')

ax.set_title('Utility, $u(x_1,x_2)$')
ax.set_xlabel('$x_1$')
ax.set_ylabel('$u(x_1,x_2)$')

ax.legend(loc='upper left');
```

<details>
<summary><strong>Two lines and a legend</strong></summary>

Both lines use the same horizontal values but different utility values. The `label` argument stores the text for each line. `ax.legend()` then displays those labels. Without `label`, the legend has no useful entry to show.

</details>


Separately:

```python
# default list of color
colors = plt.rcParams['axes.prop_cycle'].by_key()['color']

# figure
fig = plt.figure(figsize=(12,5))
ax1 = fig.add_subplot(1,2,1)
ax2 = fig.add_subplot(1,2,2)

ax1.plot(x1_vec,u_vec,color=colors[0])
ax2.plot(x1_vec,u_vec_2,color=colors[1])

for ax,x2_now in zip([ax1, ax2],(x2,x2_alt)):
    ax.set_title(f'Utility, $u(x_1,{x2_now})$',pad=10)
    ax.set_xlabel('$x_1$')
    ax.set_ylabel('$u(x_1,x_2)$')
    ax.set_ylim(0,15)
    ax.set_yticks(np.arange(0,15+1,3))

fig.suptitle('Utility values')
fig.tight_layout() # prettier
```

Many together:

```python
fig = plt.figure()
ax = fig.add_subplot(1,1,1)

n = 10
for x2 in range(1,n):
    u_vec_x2 = u_func(x1_vec,x2)
    ax.plot(x1_vec,u_vec_x2,label=f'$x_{2} = {x2}$')

ax.set_title('Utility, $u(x_1,x_2)$')
ax.set_xlabel('$x_1$')
ax.set_ylabel('$u(x_1,x_2)$')
ax.legend(loc='upper left',ncol=3,prop={'size':9});
```

## 2. <a id='toc2_'></a>[Histograms and scatter plots](#toc0_)


Draw some random noise:

```python
rng = np.random.default_rng(12345)  # for reproducibility
N = 500
x1 = 3
x2 = 2
x1_noise = np.exp(rng.normal(0,0.1,N))
x2_noise = np.exp(rng.normal(0,0.2,N))
```

Plot histogram:

```python
fig = plt.figure()
ax = fig.add_subplot(1,1,1)

ax.hist(x1_noise,bins=25,alpha=0.5,label='noise in $x_1$')
ax.hist(x2_noise,bins=25,alpha=0.5,label='noise in $x_2$');

# labels
ax.set_title('Histogram of Noisy Data',pad=10)
ax.set_xlabel('value')
ax.set_ylabel('frequency')

# legend
ax.legend();
```

```python
fig = plt.figure()
ax = fig.add_subplot(1,1,1)
ax.scatter(x1_noise,x2_noise,s=10)
ax.set_title('Scatter plot of noisy data',pad=10)
ax.set_xlabel('noise in $x_1$')
ax.set_ylabel('noise in $x_2$');
```

Calculate noisy utility measure:

```python
x1_noise = x1 + x1_noise
x2_noise = x2 + x2_noise
u_noise = u_func(x1_noise,x2_noise)
```

Scatter plot with utility as colors

```python
fig = plt.figure()
ax = fig.add_subplot(1,1,1)

h = ax.scatter(x1_noise,x2_noise,s=10,c=u_noise)
fig.colorbar(h)

# labels
ax.set_title('Scatter plot of utility in noisy data',pad=10)
ax.set_xlabel('$x_1$')
ax.set_ylabel('$x_2$');
```

<details>
<summary><strong>Choosing among common plot types</strong></summary>

- A line plot shows how one ordered variable changes with another.
- A histogram shows the distribution of one variable.
- A scatter plot shows the relationship between two variables.
- Passing `c=u_noise` maps a third variable to point color, and `fig.colorbar(h)` explains that mapping.

Choose the plot type from the question you want the reader to answer.

</details>

### 5.1. <a id='toc5_1_'></a>[Detailed formatting](#toc0_)


**Reference example:** The next cell contains many formatting options. Do not copy the whole cell into a first plot. Read it in four groups: data line, reference lines, labels and limits, then annotation and legend.


```python
fig = plt.figure(figsize=(6,4)) # width x height in inches
ax = fig.add_subplot(1,1,1)

ax.plot(x1_vec,u_vec,ls='-',lw='4',label='$x_{2}=1$')

# lines
ax.axvline(25,color='black',lw=5,alpha=0.15,zorder=0)
ax.axhline(u_func(25,x2),color='black',lw=1,zorder=0)

# labels
ax.set_title('Utility, $u(x_1,x_2)$',pad=20)
ax.set_xlabel('$x_1$')
ax.set_ylabel('$u(x_1,x_2)$')

ax.tick_params(axis='both',labelsize=12)
ax.tick_params(axis='x',labelcolor='red')

# lims
print('xlim:',ax.get_xlim())
ax.set_xlim([-1,101])
ax.set_xticks(np.arange(0,100+1,20))
print('ylim:',ax.get_ylim())
ax.set_ylim([-1,15])
ax.set_yticks(np.arange(0,14+1,2))

# annotate
ax.annotate('$u(x_1,x_2)$',xy=(25,u_func(25,x2)),xytext=(40,10),
          arrowprops=dict(facecolor='black',width=1,headwidth=8,headlength=8,shrink=0.05),
          fontsize=12,backgroundcolor='white',color='black')

# legend
ax.legend(loc='upper left'); # activate legend (uses the labels in ax.plot())
```

### 5.3. <a id='toc5_3_'></a>[3D plots](#toc0_)


#### Calculate a utility function on a grid


A 3D surface requires values on a two-dimensional grid. The next cells show both a nested-loop construction and `np.meshgrid`. This material is optional and is included as a reference for later model work.


```python
# a. settings
N = 100 # number of elements
x_max = 10 # maximum value

# b. allocate 2d numpy arrays for x1 and x2
shape_tuple = (N,N)
x1_values = np.empty(shape_tuple) # allocate 2d numpy array with shape=(N,N)
x2_values = np.empty(shape_tuple)
u_values = np.empty(shape_tuple)

# c. fill numpy arrays using a nested loop
for i in range(N): # 0,1,...,N-1
    for j in range(N): # 0,1,...,N-1
        x1_values[i,j] = (i/(N-1))*x_max # in [0,x_max]
        x2_values[i,j] = (j/(N-1))*x_max # in [0,x_max]
        u_values[i,j] = u_func(x1_values[i,j],x2_values[i,j],alpha=0.25)
```

<details>
<summary><strong>How the nested loop fills a grid</strong></summary>

The two loops visit every pair of row and column positions. For each pair `(i,j)`, the code stores one $x_1$ value, one $x_2$ value and the utility of that pair. All three arrays have shape `(N,N)` so corresponding positions describe the same point on the grid.

</details>


**Note:** x1_values and x2_values are **both 2d**. 

This is not strictly necessary in the present example. One-dimensional arrays would suffice in the nested loop. Two-dimensional arrays are used because surface and contour plots are commonly built from mesh grids.


**Alternatively:** Use the corresponding numpy function to create your grid:

```python
# a. Create numbers distributed evenly between 0 and x_max
x_vec = np.linspace(0,x_max,N)

# b. Create meshgrids similar to x1_values and x2_values above
x1_mesh,x2_mesh = np.meshgrid(x_vec,x_vec,indexing='ij')

print('Dimension of grid over x1 (same as x2): ', x1_mesh.shape) # The grid is 2d; u_func evaluates it element by element. 
u_values_mesh = u_func(x1_mesh, x2_mesh, alpha=0.25)
```

<details>
<summary><strong>`meshgrid` creates the same coordinate structure</strong></summary>

`np.meshgrid` builds two coordinate arrays from one-dimensional vectors. One array repeats the $x_1$ values in one direction and the other repeats the $x_2$ values in the other direction. Because NumPy operations work element by element, `u_func` can evaluate the full grid without an explicit loop.

</details>


Test whether the results are the same:

```python
# a. maximum absolute difference
max_abs_diff = np.max(np.abs(u_values-u_values_mesh))
print(max_abs_diff) # very close to zero     

# b. test if all values are "close"
print(np.allclose(u_values,u_values_mesh))
```

#### Plot the utility function


The first cell draws the surface with the smallest useful 3D pattern. The next cell adds a colormap, labels and a viewing angle. Treat both as reference examples.


```python
fig = plt.figure()
ax = fig.add_subplot(1,1,1,projection='3d') # 3d axis 
ax.plot_surface(x1_values,x2_values,u_values);
```

**Optional 3D formatting:** Add a colormap, labels and a viewing angle.


```python
from matplotlib import cm # for colormaps

# a. actual plot
fig = plt.figure()
ax = fig.add_subplot(1,1,1,projection='3d')
h = ax.plot_surface(x1_values,x2_values,u_values,cmap=cm.jet)
fig.colorbar(h)

# b. add labels
ax.set_title('Utility, $u(x_1,x_2)$')
ax.set_xlabel('$x_1$')
ax.set_ylabel('$x_2$')
ax.set_zlabel('$u$')

# pad z
ax.zaxis.labelpad = 0.1

# c. change view angle and zoom
ax.view_init(azim=-135,elev=30) # azim is azimuthal angle, elev is elevation angle
ax.set_box_aspect([4,4,3],zoom=0.9)

fig.tight_layout(pad=0.1)
```
