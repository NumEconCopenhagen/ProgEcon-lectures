Before we write any Python, it is worth knowing **what the machine in front of you actually is**, and what the words we will use all semester mean. None of this is needed to run a cell - but it explains why the code behaves as it does, and it makes the error messages much less frightening.

## 1. <a id='toc1_'></a>[What is a computer?](#toc0_)

A computer is a machine that can do a handful of **extremely simple things extremely fast**: add two numbers, compare two numbers, move a number from one place to another.

That is essentially all. Everything else - a spreadsheet, a video call, an estimated model - is built from those operations, piled up in enormous quantity.

The three parts you need to know about:

| part | what it is | speed | size | remembers? |
| --- | --- | --- | --- | --- |
| **CPU** (processor) | does the arithmetic | billions of operations per second | - | no |
| **RAM** (memory) | where the data being worked on lives | fast | small (GB) | only while the program runs |
| **disk** (SSD/harddrive) | where files live | slow | large (TB) | yes, permanently |

A useful picture: the **CPU is you**, the **RAM is your desk** - what you can reach right now - and the **disk is the filing cabinet** across the room. Fetching from the cabinet is slow, so you keep what you are working on on the desk.

**When your program creates a variable, it lives in RAM.** When you close Python, it is gone. Anything you want to keep has to be written to a file on the disk.

**Everything is numbers, and every number is bits.** A *bit* is a single 0 or 1, and a *byte* is 8 bits. Text, images and sound are all stored as numbers, and all numbers are stored in binary:

```python
import sys

x = 13
print('x =', x)
print('x in binary:', format(x,'08b')) # 8 binary digits: 8 + 4 + 1 = 13
print('x occupies', sys.getsizeof(x), 'bytes of memory')
```

## 2. <a id='toc2_'></a>[What is a program?](#toc0_)

A **program** is a list of instructions, executed one at a time, from top to bottom.

That is the whole idea. The art is in choosing the instructions, not in the machine.

The CPU only understands **machine code**: numbers that encode "add these two", "jump back three instructions". Nobody writes that by hand any more. We write something readable instead, and let another program translate it.

```python
# a program with three instructions
income = 100.0
tax = 0.4*income
print('income after tax:', income-tax)
```

**The computer does exactly what you wrote** - not what you meant. It has no idea what income is, that a tax rate above 1 is odd, or that you meant to write `0.4` and not `4.0`. Checking that the instructions say what you intended is *your* job, and most of this course is about how to do that.

## 3. <a id='toc3_'></a>[What is a programming language?](#toc0_)

A **programming language** is a notation that humans can read and write, and that can be translated into machine code without ambiguity.

There are two ways of doing the translation, and the difference explains a lot about how Python feels:

| | **compiled** | **interpreted** |
| --- | --- | --- |
| languages | C, C++, Fortran, Rust | **Python**, R, MATLAB, Julia\* |
| when translated | once, before you run it | while it runs, line by line |
| speed | fast | slower |
| convenience | you must compile, and be precise about types | just run it, types sort themselves out |

\*Julia compiles as it goes, so it does not fit neatly in either column.

**Python is interpreted.** That is why you can run a single cell and see the answer immediately, and why you rarely have to say what type a variable is. The price is speed: every line is translated *again* every time it runs, so a loop over a million numbers pays that cost a million times.

**Speeding up Python:**

1) Call compiled code from Python - done by a lot of packgages (e.g. `numpy`) 
1) Compile Python code with `numba` or `jax` (mentioned end-of-course)

## 4. <a id='toc4_'></a>[Why Python?](#toc0_)

1. **It is free**, and it runs on any machine.
2. **It is readable.** Python code looks close to pseudo-code, which matters when you have to understand what you did three months ago.
3. **The libraries.** `numpy` for arrays, `scipy` for optimization and statistics, `pandas` for data, `matplotlib` for figures. Almost everything an economist needs is a few lines away.
4. **Everybody uses it** - in economics, in statistics, in machine learning. Code, data and help are easy to find.

**But the language is not the point.** What you learn here - variables, loops, functions, how to structure a model, how to check an answer - transfers directly to MATLAB, R, Julia or C++. The syntax you can look up.

## 5. <a id='toc5_'></a>[Files, scripts and notebooks](#toc0_)

Your code lives in files on the disk, and they are all **plain text** - you could open them in Notepad.

* **`.py`** - a *script* or a *module*: pure Python code, nothing else. This is what real projects are written in.
* **`.ipynb`** - a *Jupyter notebook* (what you are reading): text cells, code cells **and** the output of the code, all in one file.
* **a module** is simply a `.py` file that you `import` to use its functions somewhere else.

**Rule of thumb for the projects:** notebooks are for *presenting* - text, results, figures. Put the actual machinery in a `.py` file and import it. A notebook with 400 lines of code in one cell is unreadable and impossible to reuse.

The file *extension* (`.py`, `.ipynb`, `.csv`) is just part of the name. It tells the operating system which program to open it with - it does not change what is inside.

## 6. <a id='toc6_'></a>[What is an editor?](#toc0_)

An **editor** is a program for writing plain text. It is *not* a word processor: Word stores fonts, colors and margins along with your words, and a Python file must contain nothing but the code itself.

A modern editor does much more than store characters:

* **syntax highlighting** - colors that show at a glance what is a string, a number, a comment
* **autocomplete** - suggestions while you type, and the documentation of a function when you hover over it
* **error underlining** - it marks the mistake before you run anything
* **a debugger** - stop the program mid-run and look at every variable
* **git integration** - keep track of what you changed, and when

An editor with all of that bundled together is usually called an **IDE** (*integrated development environment*). We use **VS Code**, which is free and works with any language. See the *Introduction* notebook of lecture 1 for the tour.

## 7. <a id='toc7_'></a>[The interpreter and the kernel](#toc0_)

The **Python interpreter** is itself just a program - `python.exe` on your machine. It reads your code and carries out the instructions.

When you open a notebook, VS Code starts an interpreter in the background and keeps it alive. That running process is called the **kernel**. Every cell you run is sent to the *same* kernel, which is why a variable defined in one cell still exists in the next.

```python
counter = 0
print('counter =', counter)
```

```python
counter += 1 # run this cell a few times in a row and watch
print('counter =', counter)
```

**This is the single most common source of confusion in notebooks.** The kernel remembers what you have *run*, not what you can *see*:

* running the same cell twice is not the same as running it once
* a variable can exist even though you deleted the cell that created it
* a cell far down the notebook may have been run *before* one above it

The number in `[ ]` to the left of each cell tells you the order things actually happened in.

**The cure:** when something behaves strangely, use **Restart Kernel** and then **Run All**. If the notebook does not work top to bottom in a fresh kernel, it does not work - and that is exactly how the person grading your project will run it.

## 8. <a id='toc8_'></a>[Packages and environments](#toc0_)

Python by itself is small. A **package** (or *library*) is code somebody else wrote that you can use:

* the **standard library** comes with Python: `math`, `time`, `sys`, `random`, ...
* **third-party packages** are installed separately: `numpy`, `scipy`, `pandas`, `matplotlib`, ...

You get access to a package by **importing** it:

```python
import sys
import numpy as np # 'as np' gives it a short name, used by everybody

print('Python version:', sys.version.split('|')[0].strip())
print('numpy version :', np.__version__)
print('numpy lives in:', np.__file__)
```

**Anaconda**, which you installed for this course, is Python bundled with several hundred of these packages and the tools to manage them. When you need something it does not have, you install it once with `%pip install packagename` in a cell (see the *Introduction* notebook).

## 9. <a id='toc9_'></a>[Three kinds of error](#toc0_)

Things will go wrong constantly - for everybody, at every level of experience. Errors come in three kinds, and they are *not* equally bad.

**1. Syntax errors.** The code is not valid Python, so it cannot even be read. Nothing runs. The editor usually underlines it before you press anything.

```python
# 'compile' checks the code without running it, so the notebook survives
try:
    compile('x = = 1','<my code>','exec')
except SyntaxError as e:
    print('SyntaxError:', e)
```

**2. Runtime errors.** The code is valid and starts running, but hits something impossible and stops. You get a **traceback**: read it from the **bottom up** - the last line says what went wrong, the lines above say where.

```python
try:
    income = 100
    people = 0
    print(income/people)
except Exception as e:
    print(type(e).__name__,'-',e)
```

**3. Wrong answers.** The code runs, prints a number, and the number is wrong. **No error is raised at all.**

```python
xs = [1,2,3,4]

mean_wrong = xs[0]+xs[1]+xs[2]+xs[3]/4   # forgot the parentheses
mean_right = (xs[0]+xs[1]+xs[2]+xs[3])/4

print('mean_wrong =', mean_wrong)
print('mean_right =', mean_right)
```

Both lines run happily. Only one is the average, and Python has no opinion about which one you wanted.

**The first two kinds are annoying. The third kind is dangerous**.

## 10. <a id='toc10_'></a>[Summary](#toc0_)

The words we will use from now on:

1. **CPU, RAM, disk** - the calculator, the desk and the filing cabinet. Variables live in RAM and disappear; files on the disk survive.
2. **Bit and byte** - everything is 0s and 1s, in finite quantity, which is why decimal numbers are only approximate.
3. **Program** - a list of instructions, run from top to bottom, doing exactly what you wrote.
4. **Programming language** - readable notation, translated to machine code; **compiled** once and fast, or **interpreted** line by line and flexible. Python is interpreted, so heavy work goes to libraries like `numpy`.
5. **Script (`.py`), module, notebook (`.ipynb`)** - all plain text files. Machinery in modules, presentation in notebooks.
6. **Editor / IDE** - VS Code: highlighting, autocomplete, debugger, git.
7. **Interpreter and kernel** - the kernel remembers what you *ran*, not what you *see*. Restart and Run All before you trust a notebook.
8. **Package and environment** - `import` to use, and the environment decides which version you get.
9. **Three kinds of error** - syntax, runtime and *wrong answer*. Only the last one is quiet, and it is the one that matters.
